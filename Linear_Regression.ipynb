# Linear Regression from Scratch


```python
import numpy as np
import matplotlib.pyplot as  plt
%matplotlib inline
```

```python
# --- 1. Create Synthetic  Data ---
print("Step 1: Creating synthetic data...")
```

```python
# Create 100 random 'X' values (our feature) between 0 and 2.
X = 2 * np.random.rand(100, 1)
```

```python
# Create 'y' values (our target)  based on a known line: y = 3x + 4
# We add 'np.random.randn' to add some "noise", so the points aren't perfectly linear.
# Our model's goal will be to discover the '3' (slope) and '4' (intercept).
y = 4 + 3 * X + np.random.randn(100, 1)
```

```python
print("Data created successfully.")
```

```python
# --- 2. Set Hyperparameters and Initialize Parameters ---
print("Step 2: Initializing parameters...")
```

```python
# The 'learning_rate' (alpha) controls how big of a step we take during optimization.
learning_rate = 0.01
```

```python
# 'epochs' is the number of times we will iterate over the entire dataset.
epochs = 1000
```

```python
# 'n' is the total number of data points (we use float for division later).
n = float(len(X))
```

```python
# We initialize our model's parameters, 'm' (slope) and 'b' (intercept), to zero.
# Our goal is to find the best 'm' and 'b' for our data.
m = 0.0
b = 0.0
```

```python
# We'll create a list to store the 'cost' (error) at each epoch to see if it's improving.
cost_history = []
```

```python
print(f"Starting parameters: m={m}, b={b}")
print(f"Hyperparameters: Learning Rate={learning_rate}, Epochs={epochs}")
```

```python
# --- 3. Run Gradient Descent ---
print("Step 3: Running Gradient Descent...")
```

```python
# This loop runs 'epochs' (1000) times.
for epoch in range(epochs):
    
    # --- 3a. Make a Prediction ---
    # This is the line equation: y_pred = m*X + b
    y_pred = m * X + b
    
    # --- 3b. Calculate the Cost (Error) ---
    # This is the Mean Squared Error (MSE) cost function.
    cost = (1/n) * np.sum((y_pred - y)**2)
    
    # Store the cost for this epoch so we can plot it later.
    cost_history.append(cost)
    
    # --- 3c. Calculate the Gradients ---
    # 'D_m' is the gradient (partial derivative) of the cost with respect to 'm'.
    # It tells us which direction to move 'm' to reduce the cost.
    D_m = (2/n) * np.sum(X * (y_pred - y))
    
    # 'D_b' is the gradient (partial derivative) of the cost with respect to 'b'.
    # It tells us which direction to move 'b' to reduce the cost.
    D_b = (2/n) * np.sum(y_pred - y)
    
    # --- 3d. Update the Parameters ---
    # We update 'm' by taking a small step in the *opposite* direction of its gradient.
    m = m - learning_rate * D_m
    
    # We update 'b' by taking a small step in the *opposite* direction of its gradient.
    b = b - learning_rate * D_b
    
    # Optional: Print the cost every 100 epochs to see progress.
    if (epoch + 1) % 100 == 0:
        print(f"Epoch {epoch+1}/{epochs} | Cost: {cost:.4f} | m: {m:.4f} | b: {b:.4f}")
```

```python
print("Gradient Descent finished.")
```

```python
# --- 4.  Print Final Parameters ---
print("\n--- Final Results ---")
print(f"Original parameters were: m=3, b=4")
print(f"Parameters found by model: m={m:.4f}, b={b:.4f}")
```

```python
# --- 5. Plot the Results ---
print("\nStep 4: Plotting results...")
```

```python
# Plot the 'cost_history' list to see if our cost went down.
plt.plot(cost_history)
plt.title('Cost Function (MSE) over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Cost (Error)')
```

```python
# Plot the original data points as a scatter plot.
plt.scatter(X, y, label='Original Data') 
```

```python
# Plot the final line our model learned (using the final 'm' and 'b').
plt.plot(X, m * X + b, color='red', label='Fitted Line')
plt.title('Final Model Fit')
plt.xlabel('X (Feature)')
plt.ylabel('y (Target)')
plt.legend()
```
