```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------
# Step 2: Generate Synthetic  Data
# ---------------------------------------------
np.random.seed(42)  # for reproducibility

# y = 3x + 5 + noise
X  =  np.linspace(0, 10, 100)
y  =  3 * X + 5 + np.random.randn(100) * 2  # random noise added

plt.scatter(X, y, color='blue')
plt.title('Synthetic Data (y = 3x + 5 + noise)')
plt.xlabel('X')
plt.ylabel('y')
plt.show()
```

```python
class LinearRegressionScratch:
    def __init__(self, learning_rate=0.01, epochs=1000):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weight = 0
        self.bias = 0

    def fit(self, X, y):
        n = len(X)
        for _ in range(self.epochs):
            y_pred = self.weight * X + self.bias

            # Calculate gradients
            dw = (-2/n) * np.sum(X * (y - y_pred))
            db = (-2/n) * np.sum(y - y_pred)

            # Update parameters
            self.weight -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

    def predict(self, X):
        return self.weight * X + self.bias
```

```python
model = LinearRegressionScratch(learning_rate=0.01, epochs=1000)
model.fit(X, y)
y_pred = model.predict(X)

print(f'Learned Weight (slope): {model.weight:.3f}')
print(f'Learned Bias (intercept): {model.bias:.3f}')

mse = np.mean((y - y_pred)**2)
print(f'Mean Squared Error: {mse:.3f}')

plt.scatter(X, y, color='blue', label='Actual Data')
plt.plot(X, y_pred, color='red', label='Fitted Line')
plt.title('Linear Regression From Scratch')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()
```
